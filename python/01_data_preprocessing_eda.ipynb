```python
# Step 6: Cast 'date' column to datetime data type
df['date'] = pd.to_datetime(df['date'])
```

```python
# Step 7: Feature Engineering - Financial Metrics (Revenue, Cost, Profit, Margin)
df['total_revenue'] = df['units_sold'] * df['unit_price']
df['total_cost'] = df['units_sold'] * df['unit_cost']
df['profit'] = df['total_revenue'] - df['total_cost']
df['profit_margin'] = (df['profit'] / df['total_revenue']).fillna(0)
```

```python
# Step 8: Feature Engineering - Reorder Risk & Lost Revenue Calculations
df['below_reorder_flag'] = (df['inventory_level'] <= df['reorder_point']).astype(int)
df['lost_sales_units'] = np.where(df['stockout_flag'] == 1, np.maximum(0, df['demand_forecast'] - df['units_sold']), 0)
df['lost_revenue'] = df['lost_sales_units'] * df['unit_price']
```

```python
# Step 9: Feature Engineering - Demand Forecast Error & Absolute Error
df['forecast_error'] = df['units_sold'] - df['demand_forecast']
df['abs_forecast_error'] = np.abs(df['forecast_error'])
```

```python
# Step 10: Export processed 24-column dataset to CSV
df.to_csv("cleaned_supply_chain_dataset1" , index=False)
```
